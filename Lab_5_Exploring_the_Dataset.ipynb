```python
#Load the dataset
#Read Data
#We utilize the pandas.read_csv() function for reading CSV files. However, in this version of the lab, which operates on JupyterLite, the dataset needs to be downloaded to the interface using the provided code below.
import pandas as pd
import chardet

# Step 1: Detect file encoding (optional)
with open("dataset.csv", "rb") as f:
    result = chardet.detect(f.read())
encoding = result['encoding']
print(f"Detected encoding: {encoding}")

# Step 2: Read CSV safely
try:
    df = pd.read_csv(
        "dataset.csv",
        sep=None,            # let pandas sniff the separator
        engine='python',     # Python engine is more flexible
        encoding=encoding,
        on_bad_lines='skip'  # skip malformed rows
    )
except Exception as e:
    print("Error reading CSV:", e)
    df = None

# Step 3: Inspect the dataframe
if df is not None:
    print("Dataframe shape:", df.shape)
    print(df.head())
    print("Missing values per column:")
    print(df.isna().sum())
else:
    print("Dataframe could not be loaded.")
```

```python
#Explore the dataset
#It is a good idea to print the top 5 rows of the dataset to get a feel of how the dataset will look.
import pandas as pd
import requests

# Step 1: Define file URL and local filename
file_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
file_name = "survey_data.csv"

# Step 2: Download the file
response = requests.get(file_url)
with open(file_name, "wb") as f:
    f.write(response.content)
print(f"{file_name} downloaded successfully.")

# Step 3: Load CSV safely
df = pd.read_csv(file_name, on_bad_lines='skip', engine='python')
print("CSV loaded successfully.\n")

# Step 4: Quick exploration
print("Top 5 rows of the dataset:")
print(df.head())
```

```python
#Find out the number of rows and columns
#Print the number of rows in the dataset.
num_rows = df.shape[0]
print(f"The dataset has {num_rows} rows.")
```

```python
#Print the number of columns in the dataset.
num_cols = df.shape[1]
print(f"The dataset has {num_cols} columns.")
```

```python
#Print the datatype of all columns.
print(df.dtypes)
```

```python
#Print the mean age of the survey participants.
# Define a mapping of age ranges to numeric midpoints
age_mapping = {
    "Under 18 years old": 17,
    "18-24 years old": 21,
    "25-34 years old": 29.5,
    "35-44 years old": 39.5,
    "45-54 years old": 49.5,
    "55-64 years old": 59.5,
    "65 years or older": 70,
    "Prefer not to say": None
}

# Map the Age column to numeric values
df['Age_numeric'] = df['Age'].map(age_mapping)

# Calculate the mean age
mean_age = df['Age_numeric'].mean()
print(f"The mean age of the survey participants is {mean_age:.2f}")
```

```python
#The dataset is the result of a world wide survey. Print how many unique countries are there in the Country column.
# Number of unique countries
num_countries = df['Country'].nunique()
print(f"The dataset contains {num_countries} unique countries.")
```
