```python
#Download the webpage at the url
import requests

url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
data = requests.get(url).text
```

```python
#Create a soup object
from bs4 import BeautifulSoup

soup = BeautifulSoup(data, 'html.parser')
```

```python
#Scrape the Language name and annual average salary.
# Step 3: Scrape the data
# Find the table in the HTML
table = soup.find('table')

# Create a list to store the scraped data
language_salaries = []

# Find all the rows in the table
for row in table.find_all('tr'):
    # Get all the columns in the row
    cols = row.find_all('td')
    # Make sure the row has data (not just a header row)
    if len(cols) > 0:
        # Extract the language and salary from the appropriate columns
        language = cols[1].get_text()
        salary = cols[3].get_text()
        language_salaries.append((language, salary))

# Print the scraped data
for language, salary in language_salaries:
    print(f"Language: {language}, Average Annual Salary: {salary}")
```

```python
#Save the scrapped data into a file named popular-languages.csv
import requests
from bs4 import BeautifulSoup
import csv

# Step 1: Download the webpage content
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
data = requests.get(url).text

# Step 2: Create a BeautifulSoup object
soup = BeautifulSoup(data, 'html.parser')

# Step 3: Scrape the data
language_salaries = []
table = soup.find('table')
for row in table.find_all('tr'):
    cols = row.find_all('td')
    if len(cols) > 0:
        language = cols[1].get_text()
        salary = cols[3].get_text()
        language_salaries.append([language, salary])

# Step 4: Save the data to a CSV file
csv_file_path = 'popular-languages.csv'

with open(csv_file_path, 'w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    # Write the header row
    writer.writerow(['Language', 'Average Annual Salary'])
    # Write the data rows
    writer.writerows(language_salaries)

print(f"Data saved to {csv_file_path}")
```
